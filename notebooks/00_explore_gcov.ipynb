# 00 · Explore a NISAR GCOV window

Opens a window written by `pipeline/02_search_download.py` (`data/interim/nisar/<id>/window_<date>.npz`) and shows HH, HV, the GCOV mask and their histograms. Dataset paths and mask semantics are documented in `contracts/gcov_notes.md`.

```python
import sys, json
from pathlib import Path
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT))
import numpy as np, pandas as pd, matplotlib.pyplot as plt
from pipeline.utils import sar, geo
```

```python
files = sorted((ROOT / 'data/interim/nisar').glob('*/window_*.npz'))
files[:5]
```

```python
d = np.load(files[0])
print(d.files, d['hh_db'].shape, int(d['epsg']), str(d['scene']))
print('mask values:', np.unique(d['mask']))
```

```python
fig, ax = plt.subplots(1, 3, figsize=(15, 4))
ax[0].imshow(d['hh_db'], cmap='gray', vmin=-30, vmax=0); ax[0].set_title('HH (dB)')
ax[1].imshow(d['hv_db'], cmap='gray', vmin=-35, vmax=-5); ax[1].set_title('HV (dB)')
ax[2].imshow(d['mask'], interpolation='nearest'); ax[2].set_title('GCOV mask (subswath)')
for a in ax: a.axis('off')
```

```python
for k, rng in [('hh_db', (-35, 0)), ('hv_db', (-40, -5))]:
    v = d[k][np.isfinite(d[k])]
    plt.hist(v, bins=120, range=rng, alpha=.6, label=k)
plt.legend(); plt.xlabel('dB');
```
